**Strings- sequential data: you can run loops on strings**

```python
greeting = "Welcome to the session. PYTHON Session starts at 12 : 30 PM"
print(greeting)

```

```python
greeting.split()
```

```python
title = []
lowercase = []
uppercase = []
digits = []
others = []
```

**Let's separate the words present in greeting string according to their type(lowercase,uppercase,digits).
Save the data in separate lists according to their type and display them at the end**

```python
for i in greeting.split():
    if i.istitle():
        title.append(i)
    if i.islower():
        lowercase.append(i)
    if i.isupper():
        uppercase.append(i)
    if i.isdigit():
        digits.append(i)
    else:
        others.append('Invalid Data')


print(f"Title{title},Lower{lowercase},Upper{uppercase},Digits{digits},Others{others}")
```

```python
print(greeting)
```

**Create a list that contains only even numbers from a range of numbers**

consider range of numbers between 560009 to 678456

    even numbers : 2,4,6,8,10
    odd numbers : 1,3,5,7,9,11

even numbers are all divisible by 2 and the remainder will be 0

```python
# Define an empty new list
even_nums = []
# Intialize i to 560009
i = 560009
# condition in while
while i<=678456:
    if i%2==0:
        even_nums.append(i)
    i+=1
```

**break,continue and pass**
    
    break : stops the iteration and cursor comes out of the loop
    continue: skips the current iteration
    pass: Pass statement is a null operation which is used as a placeholder for future code. It does nothing when executed.

```python
# Break scenario
# Run a loop on employee id's tuple/list until 190 id is found. Once found, the iteration should stop
emp_ids = list(range(100,200))
```

```python
for i in emp_ids:
    print(i)
    if i==190:
        break
```
